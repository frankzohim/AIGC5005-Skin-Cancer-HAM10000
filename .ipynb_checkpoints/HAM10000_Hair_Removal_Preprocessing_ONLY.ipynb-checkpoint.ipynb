```python
# CELL 4 – Hair removal function
def remove_hair(img):
    gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
    kernel = cv2.getStructuringElement(1, (17,17))
    blackhat = cv2.morphologyEx(gray, cv2.MORPH_BLACKHAT, kernel)
    _, thresh = cv2.threshold(blackhat, 10, 255, cv2.THRESH_BINARY)
    return cv2.inpaint(img, thresh, 3, cv2.INPAINT_TELEA)

print("DullRazor hair removal ready")
```

```python
# CELL 5 – ONE-TIME PREPROCESSING (run this once only!)
df = pd.read_csv('HAM10000_metadata.csv')
df['path'] = df['image_id'].apply(lambda x: f"all_images/{x}.jpg")

CLEAN_FOLDER = '/content/drive/MyDrive/HAM10000_CLEANED_WITH_HAIR_REMOVAL'
os.makedirs(CLEAN_FOLDER, exist_ok=True)

print("Starting one-time hair removal for all 10,015 images... (10–12 minutes)")
for idx in tqdm(range(len(df)), desc="Removing hair"):
    row = df.iloc[idx]
    img = cv2.cvtColor(cv2.imread(row['path']), cv2.COLOR_BGR2RGB)
    clean_img = remove_hair(img)
    cv2.imwrite(f"{CLEAN_FOLDER}/{row['image_id']}.jpg",
                cv2.cvtColor(clean_img, cv2.COLOR_RGB2BGR))

print(f"\nFINISHED! All clean images saved to:\n{CLEAN_FOLDER}")
print("You can now close this notebook forever")
```
